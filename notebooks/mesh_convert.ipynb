Converting .su2 mesh to .msh by meshio

```python
# library

import meshio
import numpy as np
```

```python
# read .su2

mesh = meshio.read("mesh_cylinder_lam.su2")

# inspect
print(mesh)                          # shows points, cells, tags
print(mesh.points.shape)             # (N, 3)
print(mesh.cells_dict.keys())        # cell types present
print(mesh.field_data)               # boundary tags / physical groups
```

```python
# add z-column to mesh
# check current shape
print(mesh.points.shape)   # (N, 2) — missing z

# add z=0 column
if mesh.points.shape[1] == 2:
    mesh.points = np.hstack([mesh.points, np.zeros((len(mesh.points), 1))])

print(mesh.points.shape)   # (N, 3) — fixed
```

```python
# meshio.write("cylinder_su2.msh", mesh, file_format="gmsh22")   # default Gmsh 4.1
meshio.write("cylinder_su2_v2.cgns", mesh)
```
